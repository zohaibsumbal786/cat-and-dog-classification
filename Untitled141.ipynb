```python
# Import necessary libraries
import os
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.preprocessing import image_dataset_from_directory
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.applications import VGG16
from PIL import Image
import warnings

# Suppress warnings related to truncated files
warnings.filterwarnings("ignore", category=UserWarning, module="PIL.TiffImagePlugin")

# Set dataset paths (Manually specified)
dataset_path = r"C:\Users\zohai\OneDrive\Desktop\kagglecatsanddogs_5340\dataset"
train_dir = os.path.join(dataset_path, "train")
validation_dir = os.path.join(dataset_path, "validation")

# Function to remove non-image, corrupt, and invalid images
def clean_dataset(directory):
    valid_extensions = (".jpg", ".jpeg", ".png")

    for category in ["train/cats", "train/dogs", "validation/cats", "validation/dogs"]:
        folder_path = os.path.join(directory, category)
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                file_path = os.path.join(folder_path, filename)

                # Remove non-image files
                if not filename.lower().endswith(valid_extensions):
                    print(f"🗑 Removing non-image file: {file_path}")
                    os.remove(file_path)
                    continue

                # Remove corrupt or problematic images
                try:
                    with Image.open(file_path) as img:
                        img.load()  # Ensure the image is fully loaded
                        img = img.convert("RGB")  # Convert to RGB
                        img.save(file_path)  # Save the corrected image
                except (IOError, OSError, AttributeError):
                    print(f"🗑 Removing corrupt/invalid image: {file_path}")
                    os.remove(file_path)

# Clean dataset before training
clean_dataset(dataset_path)

# Load datasets using `image_dataset_from_directory`
train_dataset = image_dataset_from_directory(
    train_dir,
    image_size=(150, 150),
    batch_size=32,
    label_mode="binary"
)

validation_dataset = image_dataset_from_directory(
    validation_dir,
    image_size=(150, 150),
    batch_size=32,
    label_mode="binary"
)

# ------------------------------
# 1st Model: CNN (Custom Model)
# ------------------------------
cnn_model = Sequential([
    Input(shape=(150, 150, 3)),  # Explicitly define input layer
    Conv2D(32, (3,3), activation="relu"),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation="relu"),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

cnn_model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

# Train CNN Model
history_cnn = cnn_model.fit(train_dataset, epochs=10, validation_data=validation_dataset)

# Save CNN Model
cnn_model.save("cnn_model.h5")
print("✅ CNN model saved as 'cnn_model.h5'")

# ------------------------------
# 2nd Model: VGG16 Transfer Learning
# ------------------------------
base_model = VGG16(weights="imagenet", include_top=False, input_shape=(150, 150, 3))
base_model.trainable = False  # Freeze base model layers for faster training

vgg_model = Sequential([
    base_model,
    Flatten(),
    Dense(256, activation="relu"),
    Dense(1, activation="sigmoid")
])

vgg_model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

# Train VGG16 Model
history_vgg = vgg_model.fit(train_dataset, epochs=5, validation_data=validation_dataset)

# Save VGG16 Model
vgg_model.save("vgg16_model.h5")
print("✅ VGG16 model saved as 'vgg16_model.h5'")

# ------------------------------
# Plot Accuracy & Loss Comparison
# ------------------------------
def plot_results(history, title):
    plt.plot(history.history["accuracy"], label="Training Accuracy")
    plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
    plt.title(title)
    plt.legend()
    plt.show()

# Plot CNN vs. VGG16 Model Performance
plot_results(history_cnn, "CNN Model Performance")
plot_results(history_vgg, "VGG16 Transfer Learning Model Performance")

# ------------------------------
# Compare Results and Conclusion
# ------------------------------
cnn_acc = history_cnn.history["val_accuracy"][-1]
vgg_acc = history_vgg.history["val_accuracy"][-1]

print("\n📊 **Model Performance Summary:**")
print(f"🔹 CNN Model Validation Accuracy: {cnn_acc:.4f}")
print(f"🔹 VGG16 Model Validation Accuracy: {vgg_acc:.4f}")

if vgg_acc > cnn_acc:
    print("\n✅ The **VGG16 Transfer Learning Model** performs better than the CNN model. "
          "This is expected because VGG16 is a pre-trained model with better feature extraction capabilities.")
else:
    print("\n✅ The **CNN Model** performs better than VGG16. This could be due to dataset size or hyperparameter tuning.")
```
